```python
# N-grams help us analyze the realtionship between neighboring words
# N-gram = a sequence of N tokens
# NLTK: the natural language toolkit for tokenization,stemming,lemmatization,and many other tasks
# pandas: the python library for working with structure data
# DataFrames: the spreadsheets inside python
# A DataFrame stores data in rows and columns
# matplotlib: the Python library for creating charts and visualizations
# matpotlib can take the results of our text analyzes and turn them into graphs
```

```python
import nltk
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tokens = [
    'python', 'data', 'AI', 'python', 'machine',
    'data', 'learning', 'AI', 'python', 'analytics',
    'data', 'machine', 'AI', 'python', 'learning',
    'data', 'analytics', 'AI', 'machine', 'python',
    'learning', 'data', 'AI', 'analytics', 'python',
    'machine', 'learning', 'data', 'AI', 'analytics',
    'python', 'machine', 'learning', 'data', 'AI',
    'analytics', 'python', 'machine', 'learning', 'data',
    'AI', 'analytics', 'python', 'machine', 'learning',
    'data', 'AI', 'analytics', 'python', 'machine'
]
print(tokens)
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens,1)).value_counts())
print(unigrams[:40])
```

```python
unigrams[:10].sort_values().plot.barh(color="lightsalmon", width=0.5, figsize=(12,8))
plt.title("10 Most Frequently Occuring Unigrams")
```

```python
bigrams = (pd.Series(nltk.ngrams(tokens,2)).value_counts())
print(unigrams[:40])
```

```python
bigrams[:10].sort_values().plot.barh(color="lightsalmon", width=0.5, figsize=(12,8))
plt.title("10 Most Frequently Occuring Unigrams")
```

```python
trigrams = (pd.Series(nltk.ngrams(tokens,3)).value_counts())
print(unigrams[:40])
```

```python
trigrams[:10].sort_values().plot.barh(color="lightsalmon", width=0.5, figsize=(12,8))
plt.title("10 Most Frequently Occuring Unigrams")
```
