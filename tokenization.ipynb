```python
# 'punkt_tab' : a table used to figure out where sentences begin and end
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sentences = "She tied a colorful string around the gift box. He wore a nice necklace with a string of pearls. "
```

```python
sent_tokenize(sentences)
```

```python
sentence = "He wore a nice necklace with a string of pearls."
word_tokenize(sentence)
```
