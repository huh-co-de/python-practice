```python
# Stemming: reducing words to their base form
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
connect_tokens  = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

```python
for t in connect_tokens:
    print(t, ":", ps.stem(t))
```

```python
automate_tokens = ['automating', 'automated', 'automation', 'automate', 'automates']
```

```python
for t in automate_tokens:
    print(t, ":", ps.stem(t))
```

```python
likes_tokens = ['likes','better','worse']
```

```python
for t in likes_tokens:
    print(t, ":", ps.stem(t))
```

Lemmatization



```python
# Lemmatization reduces a word to a meaningful base form while preserving its meaning
```

```python
import nltk
nltk.download('wordnet') #The built-in dictionary 
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t, ":" , lemmatizer.lemmatize(t)) #It takes the token t and looks it up in the WordNet dictionary to return its proper baseform
```
