```python
customer_reviews = [
    """sols bought this wireless speaker last week, and sol have mixed feelings. 
The sound quality is absolutely amazing for its size, and the bass is punchy! 
However, the battery life doesn't last anywhere near the advertised 12 hours—it died after just 5 hours of continuous playback.""",

    """Absolutely love this product! The shipping was lightning fast, and it arrived in perfect condition. 
It works exactly as described and has made my daily routine so much easier. 
Highly recommend to anyone on the fence!""",

    """Terrible customer service. The item arrived broken, and sols have been trying to get a refund for three days. 
Every time sols call, sol get put on hold for over twenty minutes. 
Avoid this seller at all costs."""
]
```

```python
sols_reviews = []
```

```python
patten_to_find = r"sols?"
```

```python
for string in customer_reviews:
    if(re.search(patten_to_find,string)):
        sols_reviews.append(string)
        
```

```python
print(sols_reviews)
```

```python
# r"^a" matches any string that beginns with 'a'
a_reviews = []
pattern_to_find = r"^A"
for string in customer_reviews:
    if(re.search(pattern_to_find,string)):
        a_reviews.append(string)
```

```python
print(a_reviews)
```

```python
# r"y$" matches any string that end with 'y'
y_reviews = []
pattern_to_find = r"y$"

for string in customer_reviews:
    # print(string, end="  ")
    if(re.search(pattern_to_find, string)):
            # print(string, end="  ")
            y_reviews.append(string)
```

```python
print(y_reviews)
```

```python
# "|" works like an "or"
cust_reviews = []
pattern_to_find = r"(this|the)"
for string in customer_reviews:
    if(re.search(pattern_to_find,string)):
        cust_reviews.append(string)
print(cust_reviews)
```

```python
# r"[^\w\s]" []--> a set of characters we want to match
# ^--> "not"
# \w--> word characters
# \s-->whitespace

no_punct_reviews = []
pattern_to_find = r"[^\w\s]"
```

```python
for string in customer_reviews:
    if(re.search(pattern_to_find,string)):
        no_punct_reviews.append(string)
print(no_punct_reviews)
```

```python
for string in customer_reviews:
    if(re.sub(pattern_to_find," ", string)):
        no_punct_reviews.append(string)
print(no_punct_reviews)
```
