```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words('english')
print(en_stopwords)
```

```python
sentence = """This is a classic English sentence because it is a pangram, meaning it contains every single letter of the English alphabet at least once.
Would you like a sentence that is funny, inspirational, grammatically complex, or written about a specific topic? Let me know what kind of sentence you need!"""
```

```python
sentence_no_stopwords = ' '.join(word for word in sentence.split() if word not in en_stopwords)
```

```python
print(sentence_no_stopwords)
```
